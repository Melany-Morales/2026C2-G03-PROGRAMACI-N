# Caso 1: Control de ventas diarias de una soda

## Problema
Una soda ubicada en San Pedro registra sus ventas de lunes a viernes. El programa debe calcular el total semanal, el promedio diario y clasificar el resultado según una meta.

## Entrada / datos
Use esta lista base:

```python
ventas_soda = [85000, 92000, 78000, 110000, 97000]
meta_semanal = 450000
```

## Proceso esperado

- Calcular total semanal.
- Calcular promedio diario.
- Comparar el total con la meta.
- Clasificar:
  - `Meta alcanzada` si el total es mayor o igual a la meta.
  - `Cerca de la meta` si el total alcanza al menos el 80% de la meta.
  - `Requiere atención` si no alcanza el 80%.

## Salida esperada
Debe mostrar total, promedio, porcentaje de cumplimiento y estado final.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|---|---|---|---|
| Caso normal | `[85000, 92000, 78000, 110000, 97000]`, meta `450000` | Meta alcanzada | Total=462000, Promedio=92400, Cumplimiento≈102.67%, Estado="Meta alcanzada" |
| Caso límite | total exactamente igual a la meta | Meta alcanzada | Total=450000, Promedio=90000, Cumplimiento=100%, Estado="Meta alcanzada" |
| Caso especial | total menor al 80% de la meta | Requiere atención | Total=150000, Promedio=30000, Cumplimiento≈33.33%, Estado="Requiere atención" |

````python
# Caso 1 - Escriba su solución aquí
def clasificar_estado(total, meta):
    if total >= meta:
        return "Meta alcanzada"
    elif total >= meta * 0.8:
        return "Cerca de la meta"
    else:
        return "Requiere atención"

ventas_soda = [85000, 92000, 78000, 110000, 97000]
meta_semanal = 450000

total_semanal = sum(ventas_soda)
promedio_diario = total_semanal / len(ventas_soda)
porcentaje_cumplimiento = (total_semanal / meta_semanal) * 100
estado_final = clasificar_estado(total_semanal, meta_semanal)

print("\nReporte de ventas semanales")
print("Ventas diarias:", ventas_soda)
print("Total semanal:", total_semanal)
print("Promedio diario:", promedio_diario)
print("Cumplimiento de meta:", f"{porcentaje_cumplimiento:.2f}%")
print("Estado final:", estado_final)
````

# Caso 2: Asistencia a una actividad cultural

## Problema
Una actividad cultural en Costa Rica registra asistencia por bloques horarios. El programa debe calcular la asistencia promedio y clasificar el nivel de ocupación.

## Entrada / datos
Use esta lista base:

```python
asistencias = [120, 180, 210, 250, 230]
capacidad = 300
```

## Proceso esperado

- Recorrer la lista con un ciclo `for`.
- Calcular asistencia total y promedio.
- Contar cuántos bloques superaron el 70% de capacidad.
- Clasificar el promedio de ocupación:
  - `Alta ocupación` si el promedio es mayor o igual al 80% de la capacidad.
  - `Ocupación media` si el promedio es mayor o igual al 50%.
  - `Baja ocupación` si el promedio es menor al 50%.

## Salida esperada
Debe mostrar total de asistentes, promedio, bloques sobre 70% y estado de ocupación.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|---|---|---|---|
| Caso normal | `[120, 180, 210, 250, 230]`, capacidad `300` | Ocupación media o alta según cálculo | Total=990, Promedio=198, Bloques sobre 70%=3, Ocupación≈66%, Estado="Ocupación media" |
| Caso límite | promedio exactamente 80% de capacidad | Alta ocupación | Total=1200, Promedio=240, Bloques sobre 70%=5, Ocupación=80%, Estado="Alta ocupación" |
| Caso especial | lista con asistencias muy bajas | Baja ocupación | Total=150, Promedio=30, Bloques sobre 70%=0, Ocupación=10%, Estado="Baja ocupación" |

````python
def clasificar_ocupacion(promedio, capacidad):
    porcentaje = (promedio / capacidad) * 100
    if porcentaje >= 80:
        return "Alta ocupación"
    elif porcentaje >= 50:
        return "Ocupación media"
    else:
        return "Baja ocupación"

asistencias = [120, 180, 210, 250, 230]
capacidad = 300

total_asistencias = sum(asistencias)
promedio_asistencias = total_asistencias / len(asistencias)

bloques_sobre_70 = 0
for asistencia in asistencias:
    if asistencia >= capacidad * 0.7:
        bloques_sobre_70 += 1

porcentaje_ocupacion = (promedio_asistencias / capacidad) * 100
estado_final = clasificar_ocupacion(promedio_asistencias, capacidad)

print("\n--- Reporte de asistencia ---")
print("Asistencias registradas:", asistencias)
print("Total de asistentes:", total_asistencias)
print("Promedio de asistentes:", promedio_asistencias)
print("Bloques sobre 70%:", bloques_sobre_70)
print("Porcentaje de ocupación:", f"{porcentaje_ocupacion:.2f}%")
print("Clasificación:", estado_final)
````

# Caso 3: Precios de productos de feria

## Problema
Una persona compara precios de un mismo producto en diferentes ferias del agricultor. El programa debe analizar los precios y detectar si hay mucha variación.

## Entrada / datos
Use esta lista base:

```python
producto = "Tomate"
precios = [900, 850, 1000, 950, 1200]
```

## Proceso esperado

- Calcular precio promedio.
- Identificar precio mínimo y máximo.
- Calcular rango de precios: `máximo - mínimo`.
- Clasificar la variación:
  - `Variación alta` si el rango es mayor o igual a 300.
  - `Variación moderada` si el rango es mayor o igual a 150.
  - `Variación baja` si el rango es menor a 150.
- Usar un ciclo para contar cuántos precios están por encima del promedio.

## Salida esperada
Debe mostrar producto, promedio, mínimo, máximo, rango, precios sobre promedio y clasificación.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|---|---|---|---|
| Caso normal | `[900, 850, 1000, 950, 1200]` | Variación alta | Promedio=980, Mín=850, Máx=1200, Rango=350, Precios sobre promedio=2, Estado="Variación alta"  |
| Caso límite | rango exactamente `300` | Variación alta | Promedio=950, Mín=800, Máx=1100, Rango=300, Precios sobre promedio=1, Estado="Variación alta" |
| Caso especial | todos los precios iguales | Variación baja |  Promedio=1000, Mín=1000, Máx=1000, Rango=0, Precios sobre promedio=0, Estado="Variación baja" |

````python
# Caso 3 - Escriba su solución aquí

def clasificar_variacion(rango):
    if rango >= 300:
        return "Variación alta"
    elif rango >= 150:
        return "Variación moderada"
    else:
        return "Variación baja"

producto = "Tomate"
precios = [900, 850, 1000, 950, 1200]

precio_minimo = min(precios)
precio_maximo = max(precios)
precio_promedio = sum(precios) / len(precios)
rango_precios = precio_maximo - precio_minimo
clasificacion = clasificar_variacion(rango_precios)

precios_sobre_promedio = 0
for precio in precios:
    if precio > precio_promedio:
        precios_sobre_promedio += 1

print("\n--- Reporte de precios ---")
print("Producto:", producto)
print("Precios registrados:", precios)
print("Precio mínimo:", precio_minimo)
print("Precio máximo:", precio_maximo)
print("Precio promedio:", precio_promedio)
print("Rango de precios:", rango_precios)
print("Precios sobre promedio:", precios_sobre_promedio)
print("Clasificación:", clasificacion)
````

# Caso 4: Reto integrador - Red EmprendeCR

## Problema
Una red de emprendimientos costarricenses desea analizar el desempeño semanal de varias sedes. Cada sede tiene nombre, provincia, tipo de negocio, ventas diarias y meta semanal.

Este reto integra aprendizajes de las semanas 1 a 7:

- Semana 1: comprensión del problema y algoritmo.
- Semana 2: variables y entrada-proceso-salida.
- Semana 3: decisiones con condicionales.
- Semana 4: validación y código limpio.
- Semana 5: indicadores para análisis de datos.
- Semana 6: listas y ciclos.
- Semana 7: listas de diccionarios, funciones y elección de estructuras.

## Datos base

Use esta lista de diccionarios:

```python
sedes = [
    {
        "nombre": "Soda San Pedro",
        "provincia": "San José",
        "tipo": "Alimentación",
        "ventas": [85000, 92000, 78000, 110000, 97000],
        "meta": 450000,
    },
    {
        "nombre": "Panadería Cartago",
        "provincia": "Cartago",
        "tipo": "Panadería",
        "ventas": [65000, 70000, 72000, 69000, 75000],
        "meta": 380000,
    },
    {
        "nombre": "Cafetería Liberia",
        "provincia": "Guanacaste",
        "tipo": "Cafetería",
        "ventas": [98000, 105000, 99000, 120000, 115000],
        "meta": 500000,
    },
]
```

## Requisitos mínimos

1. Crear una función `calcular_total(ventas)`.
2. Crear una función `calcular_promedio(ventas)`.
3. Crear una función `clasificar_sede(total, meta)`.
4. Recorrer todas las sedes con un ciclo `for`.
5. Imprimir un reporte por sede con:
   - nombre;
   - provincia;
   - total semanal;
   - promedio diario;
   - porcentaje de cumplimiento;
   - estado.
6. Crear un `set` con las provincias presentes.
7. Crear una lista de tuplas `(nombre, total)` para guardar el ranking base.
8. Identificar la sede con mayor venta total.

## Clasificación requerida

- `Meta alcanzada`: total mayor o igual a la meta.
- `Cerca de la meta`: total mayor o igual al 80% de la meta.
- `Requiere atención`: total menor al 80% de la meta.

## Salida esperada
El reporte debe ser claro y comprensible para una persona que no está viendo el código.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|---|---|---|---|
| Caso normal | datos base | Reporte para las 3 sedes | Soda San Pedro: Meta alcanzada; Panadería Cartago: Requiere atención; Cafetería Liberia: Cerca de la meta. Provincias={San José, Cartago, Guanacaste}. Ranking ordenado. Mejor sede=Cafetería Liberia. |
| Caso límite | una sede con total igual a la meta | Meta alcanzada | Total=450000, Promedio=90000, Cumplimiento=100%, Estado="Meta alcanzada" |
| Caso especial | una sede con ventas muy bajas | Requiere atención | Total=36000, Promedio=7200, Cumplimiento≈9%, Estado="Requiere atención" |

````python
# Caso 4 - Reto integrador
def calcular_total(ventas):
    return sum(ventas)

def calcular_promedio(ventas):
    return sum(ventas) / len(ventas)

def clasificar_sede(total, meta):
    if total >= meta:
        return "Meta alcanzada"
    elif total >= meta * 0.8:
        return "Cerca de la meta"
    else:
        return "Requiere atención"

sedes = [
    {
        "nombre": "Soda San Pedro",
        "provincia": "San José",
        "tipo": "Alimentación",
        "ventas": [85000, 92000, 78000, 110000, 97000],
        "meta": 450000,
    },
    {
        "nombre": "Panadería Cartago",
        "provincia": "Cartago",
        "tipo": "Panadería",
        "ventas": [65000, 70000, 72000, 69000, 75000],
        "meta": 380000,
    },
    {
        "nombre": "Cafetería Liberia",
        "provincia": "Guanacaste",
        "tipo": "Cafetería",
        "ventas": [98000, 105000, 99000, 120000, 115000],
        "meta": 500000,
    },
]

# Provincias presentes
provincias = set()
ranking = []

print("\n--- Reporte de sedes ---")
for sede in sedes:
    total = calcular_total(sede["ventas"])
    promedio = calcular_promedio(sede["ventas"])
    porcentaje = (total / sede["meta"]) * 100
    estado = clasificar_sede(total, sede["meta"])

    provincias.add(sede["provincia"])
    ranking.append((sede["nombre"], total))

    print(f"\nSede: {sede['nombre']} ({sede['provincia']})")
    print("Tipo:", sede["tipo"])
    print("Total semanal:", total)
    print("Promedio diario:", promedio)
    print("Cumplimiento de meta:", f"{porcentaje:.2f}%")
    print("Estado:", estado)

# Ranking y mejor sede
ranking.sort(key=lambda x: x[1], reverse=True)
mejor_sede = ranking[0]

print("\n--- Provincias presentes ---")
print(provincias)

print("\n--- Ranking de sedes ---")
print(ranking)

print("\n--- Mejor sede ---")
print(mejor_sede)
````
